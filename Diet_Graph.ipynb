```python
import matplotlib.pyplot as plt

# --- 1. Average Macro Data (Based on your 34-day logs) ---
# These values represent your average daily intake over the tracking period.
avg_protein_g = 88   # (Chicken breast, eggs, beef, pork)
avg_carbs_g = 205    # (White rice, corn rice, pao zi, fruit)
avg_fat_g = 72       # (Peanuts, fried items, duck, cooking oil)

# --- 2. Calculate Caloric Contribution ---
# Protein: 4 kcal/g | Carbs: 4 kcal/g | Fats: 9 kcal/g
cals_protein = avg_protein_g * 4
cals_carbs = avg_carbs_g * 4
cals_fat = avg_fat_g * 9
total_cals = cals_protein + cals_carbs + cals_fat

# Calculate Percentages
pct_protein = (cals_protein / total_cals) * 100
pct_carbs = (cals_carbs / total_cals) * 100
pct_fat = (cals_fat / total_cals) * 100

# --- 3. Plotting Configuration ---
labels = [
    f'PROTEIN\n({avg_protein_g}g)',
    f'CARBS\n({avg_carbs_g}g)',
    f'FATS\n({avg_fat_g}g)'
]

sizes = [cals_protein, cals_carbs, cals_fat]
colors = ['#ff6b6b', '#4ecdc4', '#ffe66d']  # Red for protein, Teal for carbs, Yellow for fats
explode = (0.1, 0, 0)  # "Explode" the protein slice to highlight it

# Create the figure
plt.figure(figsize=(10, 8), facecolor='white')

# Create the Pie Chart
patches, texts, autotexts = plt.pie(
    sizes,
    explode=explode,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    shadow=True,
    startangle=140,
    textprops={'fontsize': 12, 'fontweight': 'bold', 'color': '#2c3e50'}
)

# Style the percentage text inside the slices
for autotext in autotexts:
    autotext.set_color('black')
    autotext.set_fontsize(14)

# Chart Title and Subtitle
plt.title(f'34-Day Average Diet Composition\nTotal Avg: {int(total_cals)} kcal/day',
          fontsize=16, fontweight='bold', pad=20)

plt.axis('equal')  # Ensures the pie is a circle
plt.tight_layout()

# Save or Show
plt.savefig('my_diet_circle_graph_clean.png', dpi=300)
plt.show()
```
